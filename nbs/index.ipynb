# juno-adx-client

> Azure Data Explorer client helpers for Juno notebooks

This package wraps the small Azure authentication and query helpers that were originally sketched in the repo notebook.
It keeps the runtime dependencies explicit and moves connection setup behind functions so imports stay safe in notebooks and modules.

## Developer Guide

The package source lives in `nbs/00_core.ipynb` and exports into `juno_adx_client/core.py`.

### Install juno_adx_client in Development mode

```sh
$ pip install -e .[dev]

# export notebook changes into the package
$ nbdev_export

# run the full nbdev prep pipeline when Quarto is available
$ nbdev_prepare
```

## Usage

### Installation

Install from the repo in editable mode while developing:

```sh
$ pip install -e .[dev]
```

Runtime dependencies include:

- azure-identity
- azure-kusto-data
- azure-keyvault-secrets
- pandas
